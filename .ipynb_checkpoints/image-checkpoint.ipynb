```python
%load_ext autoreload
%autoreload 2

import matplotlib
import torch
import torch.nn as nn
import torch.nn.functional as F
import ml_collections
import sys
sys.path.insert(0, '/scratch/aa11803/tauLDR')
from config.eval.cifar10 import get_config as get_eval_config
import lib.utils.bookkeeping as bookkeeping
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt

import lib.utils.utils as utils
import lib.models.models as models
import lib.models.model_utils as model_utils
import lib.sampling.sampling as sampling
import lib.sampling.sampling_utils as sampling_utils
from tqdm import tqdm

%matplotlib inline

eval_cfg = get_eval_config()
train_cfg = bookkeeping.load_ml_collections(Path(eval_cfg.train_config_path))

for item in eval_cfg.train_config_overrides:
    utils.set_in_nested_dict(train_cfg, item[0], item[1])

S = train_cfg.data.S
device = torch.device(eval_cfg.device)

model = model_utils.create_model(train_cfg, device)

loaded_state = torch.load(Path(eval_cfg.checkpoint_path),
    map_location=device, weights_only=False)

modified_model_state = utils.remove_module_from_keys(loaded_state['model'])
model.load_state_dict(modified_model_state)

model.eval()

def imgtrans(x):
    x = np.transpose(x, (1,2,0))
    return x
```

```python
# -------------- Sample the model ------------------
num_samples = 1000
min_t = 0.001
num_steps = int(2**16)
n_states = 256
seed = 0

sampler = sampling_utils.get_sampler(eval_cfg)
samples, diffs, rr_hist = sampler.sample(model, num_samples)
np.save(f'../d3pm/save_alphas/tauldr_rr_hist_nst{num_steps}_nsa{num_samples}_s{seed}.npy', rr_hist)
```

```python
# calculate forward rates
p0 = np.load('../d3pm/save_alphas/cifar10_p0.npy')

min_rates, max_rates, av_rates = [], [], []
range_ = np.arange(n_states)
ts = np.linspace(min_t, 1, num_steps)
for t in tqdm(ts):
    pt = p0 @ model.transition(torch.tensor([t]).cuda().float())[0].cpu().numpy()
    plt.plot(pt, color=(1-t, 1-t, 1-t))
    rates = -model.rate(torch.tensor([t]).cuda().float())[0].cpu().numpy()[range_, range_]
    av_rate = rates @ pt
    av_rates.append(av_rate)
    min_rates.append(rates.min())
    max_rates.append(rates.max())
av_rates = np.array(av_rates)
np.save(f'../d3pm/save_alphas/tauldr_fwd.npy', av_rates)
```

### plot

```python
plt.figure()
plt.semilogy(diffs[::-1] * num_steps, label='empirical rates')
plt.semilogy(av_rates, label='fwd rates')
plt.semilogy(rr_hist[::-1], label="bwd rates")
plt.legend()
plt.figure()
plt.semilogy(diffs[::-1] * num_steps, label='empirical rates')
plt.semilogy(av_rates, label='fwd rates')
plt.semilogy(rr_hist[::-1], label="bwd rates")
plt.legend()
plt.ylim(100, 300)
plt.xlim(0, 500)
```

```python
plt.figure()
# plt.semilogy()
rp = rr_hist[::-1]
rt = av_rates
kls = rp - rt + rt * np.log(rt / rp)
plt.plot(ts, kls)
plt.xlabel("t")
plt.ylabel("kls")

plt.figure(figsize=[7, 3])
plt.plot(ts, 0*ts, color='black')
plt.plot(ts[:-1], (rp - rt)[:-1])
plt.ylabel("$\Delta$ Transition rate", fontsize=20)
plt.xlabel("t", fontsize=20)
plt.xticks(fontsize=15)
plt.yticks(fontsize=15)
plt.ylim(-31, 53)
plt.xlim(0, 1)

# plt.semilogy(diffs * 2**10)
# plt.semilogy(-rates_min[::-1])
# plt.semilogy(-rates_max[::-1])
```

```python
def moving_average(x, window=509):
    return torch.nn.functional.avg_pool1d(x.unsqueeze(0), window, stride=1, padding=window//2).squeeze(0)
plt.figure(figsize=[7, 2])
plt.plot(ts, 0*ts, color='black', ls='--')
plt.plot(ts[ts>0.01], moving_average(torch.tensor(rp - rt))[ts>0.01],
         lw=3, color='black')
plt.ylabel("$\Delta$ Trans. rate", fontsize=20)
plt.xlabel("t", fontsize=20)
plt.xticks([0, 1], [0, 1],fontsize=15)
plt.yticks(fontsize=15)
plt.ylim(-0.5, 1)
plt.xlim(0, 1)
# plt.tight_layout()
```
